```python
import pandas as pd

# Load data
species_data = pd.read_excel("All_plots_Desktop.xlsx", sheet_name='input')

# Trim column names (if needed)
species_data.columns = species_data.columns.str.strip()

# Find index of "Number of species"
index = species_data.columns.get_loc("Number of species")

# Select specific columns (equivalent to dplyr::select)
selected_columns = species_data.loc[:, ["Table number","Testsite"] + list(species_data.columns[index:])]

# Create "PlotIdentifier" by combining "Table number" and "Testsite"
selected_columns["PlotIdentifier"] = selected_columns["Table number"].astype(str) + "_" + selected_columns["Testsite"].astype(str)

# Reorder columns to put "PlotIdentifier" first
column_order = ["PlotIdentifier"] + [col for col in selected_columns.columns if col != "PlotIdentifier"]
selected_columns = selected_columns[column_order]
```

```python
# Drop specified columns
dataset_all_subset = selected_columns.drop(columns=["Table number", "Testsite","Number of species"], 
                                           errors='ignore')
```

```python
dataset_all_subset.set_index("PlotIdentifier", inplace=True)
```

```python
# Drop specified columns
dataset_all_subset_new = dataset_all_subset.drop(columns=["PlotIdentifier"], 
                                           errors='ignore')
```

```python
dataset_all_subset_new.fillna(0, inplace=True)
```
